Q2：**신규 가입 상품 찾기**: 고객별로 전월 대비 새로 보유하게 된 상품(0 → 1)을 찾아서, 월별·상품별 신규 가입 건수를 계산하여 전체 기간 중 신규 가입이 가장 많은 상품 5개 출력 (즉, 상품 컬럼 아이디, 예를 들어 ind_nomina_ult1)
- Pandas의 .agg, .diff 함수 사용하지 말 것

```python
df = data.copy()

# 상품
products = ['ind_ahor_fin_ult1', 'ind_aval_fin_ult1',
       'ind_cco_fin_ult1', 'ind_cder_fin_ult1', 'ind_cno_fin_ult1',
       'ind_ctju_fin_ult1', 'ind_ctma_fin_ult1', 'ind_ctop_fin_ult1',
       'ind_ctpp_fin_ult1', 'ind_deco_fin_ult1', 'ind_deme_fin_ult1',
       'ind_dela_fin_ult1', 'ind_ecue_fin_ult1', 'ind_fond_fin_ult1',
       'ind_hip_fin_ult1', 'ind_plan_fin_ult1', 'ind_pres_fin_ult1',
       'ind_reca_fin_ult1', 'ind_tjcr_fin_ult1', 'ind_valo_fin_ult1',
       'ind_viv_fin_ult1', 'ind_nomina_ult1', 'ind_nom_pens_ult1',
       'ind_recibo_ult1'] 

# 제거할 columns
del_columns = ["ind_empleado", "pais_residencia", "sexo", "age", "fecha_alta", "ind_nuevo", "indresi",
                      "antiguedad", "indrel", "indrel_1mes", "tiprel_1mes", "tiprel_1mes", "indext",
                      "conyuemp", "canal_entrada", "indfall", "tipodom", "cod_prov", "nomprov", "renta", "segmento",
                      "ind_actividad_cliente", "ult_fec_cli_1t"]

df = df.drop(columns=del_columns)
```

```python
# 데이터 프레임을 "ncopders"와 "fecha_dato"의 내림차순으로 정렬
df = df.sort_values(["ncodpers", "fecha_dato"])

# 달 리스트
months = list(df.groupby(by="fecha_dato", as_index=True).sum().index) 
new_product_list = []
# 이웃한 달끼리 상품의 신규 가입 건수 비교
for i in range(len(months)-1):
    df_prev = df.loc[df.loc[:, "fecha_dato"]==months[i], ["ncodpers"]+products].groupby(by="ncodpers", as_index=True).sum()
    df_next = df.loc[df.loc[:, "fecha_dato"]==months[i+1],["ncodpers"]+products].groupby(by="ncodpers", as_index=True).sum()
   
    new_product_df = df_next.sub(df_prev).fillna(0)
    new_product_list.append(new_product_df)
new_product_df = pd.concat(new_product_list, keys=months[1:], names=["fecha_dato", "ncodpers"])
```

```python
new_product_df
```

```python
# 지난 달과 달리 다음 달에 신규 가입한 상품만 True라고 표시한 데이터 프레임
total = new_product_df == 1 
total
```

```python
monthly_new = total.groupby(level="fecha_dato").sum()
monthly_new
```

```python
total = monthly_new.sum()
```

```python
# 가장 많은 신규 가입상품 건수 5개 출력 과정
tmp = {}
for i in products:
    tmp[i] = total[i]

result = sorted(tmp.items(), key=lambda x : x[1], reverse=True)

prod_name = []
for j in result:
    prod_name.append(j[0])

print(prod_name[:5])
```

Q3：**마지막 달** 데이터에서 가입 상품의 개수를 세어 3개 이상 혹은 3개 미만의 여부를 예측하고자 합니다. 다음의 기준 중에서 가입상품 3개 이상 여부에 대해 가장 관련성이 큰 기준을 **chi-square**를 이용해 찾으세요.
- 1) 성별 (H / V)
- 2) 나이 65이상 (senior), 65 미만 30 이상 (adult), 30 미만 (young)
- 3) segment
- 4) ind_empleado (A: 재직, B: 전직 직원, F: 계열사 소속, N: 직원 아님, P: 휴직·소극 상태)
- 5) tiprel_1mes (A: 활동, I: 비활동, P: 전 고객, R: 잠재 고객)

```python
df=data.copy()
print(df.head())
```

```python
last_month = df.loc[df['fecha_dato'] == '2016-05-28']

product_cols = [
    'ind_ahor_fin_ult1',
    'ind_aval_fin_ult1',
    'ind_cco_fin_ult1',
    'ind_cder_fin_ult1',
    'ind_cno_fin_ult1',
    'ind_ctju_fin_ult1',
    'ind_ctma_fin_ult1',
    'ind_ctop_fin_ult1',
    'ind_ctpp_fin_ult1',
    'ind_deco_fin_ult1',
    'ind_deme_fin_ult1',
    'ind_dela_fin_ult1',
    'ind_ecue_fin_ult1',
    'ind_fond_fin_ult1',
    'ind_hip_fin_ult1',
    'ind_plan_fin_ult1',
    'ind_pres_fin_ult1',
    'ind_reca_fin_ult1',
    'ind_tjcr_fin_ult1',
    'ind_valo_fin_ult1',
    'ind_viv_fin_ult1',
    'ind_nomina_ult1',
    'ind_nom_pens_ult1',
    'ind_recibo_ult1'
]

# 고객별 가입 상품 개수 계산
last_month['product_count'] = last_month[product_cols].fillna(0).sum(axis=1)

# 상품 3개 이상 / 3개 미만 분류
last_month['product_group'] = np.where(
    last_month['product_count'] >= 3,
    '3plus',
    'less3'
)

# 나이 숫자로 변환
last_month['age'] = pd.to_numeric(last_month['age'], errors='coerce')

last_month['age_group'] = 'unknown'

last_month.loc[last_month['age'] >= 65, 'age_group'] = 'senior'
last_month.loc[
    (last_month['age'] >= 30) & (last_month['age'] < 65),
    'age_group'
] = 'adult'
last_month.loc[last_month['age'] < 30, 'age_group'] = 'young'
```

```python
# 카이제곱 검정 대상 변수
features = [
    'sexo',
    'age_group',
    'segmento',
    'ind_empleado',
    'tiprel_1mes'
]

results = []

for feature in features:

    # 필요한 데이터 선택 후 결측값 제거
    sub_data = last_month[[feature, 'product_group']].dropna()

    if feature == 'age_group':
        sub_data = sub_data.loc[sub_data['age_group'] != 'unknown']

    # 교차표 만들기
    table = pd.crosstab(
        sub_data[feature],
        sub_data['product_group']
    )

    # 카이제곱 검정
    chi2, p_value, dof, expected = chi2_contingency(table)

    results.append({
        'feature': feature,
        'chi2': chi2,
        'p_value': p_value
    })


result = pd.DataFrame(results)
result = result.sort_values('chi2', ascending=False)

print('카이제곱 검정 결과')
print(result)

print('가장 관련성이 큰 기준:', result.iloc[0]['feature'])
```
